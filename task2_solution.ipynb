# Purpose

This document contains: (A) a structured set of follow-up questions for the researcher (Samuel Wiest) grouped by theme, (B) a clear protocol for conducting the elicitation interview, (C) a plan for documenting responses, (D) a preliminary list of functional and non-functional requirements (using a hybrid of User Stories + EARS), and (E) a short paragraph describing how the team will gather and refine requirements going forward.

# Understanding 

Goal: Evaluate robot localization and navigation performance in simulation and analyze correlations between scenario variations and outcomes.

Variations considered: map/layout differences, obstacles (location/size), sensor disturbances/drop rate, start/goal/waypoint variability.

Data available: logs from many TurtleBot runs in Gazebo/ROS2, scenario files, rosbag2 recordings, maps, run metadata.

Researcher interest: measures/metrics for performance, anomaly/failure detection, and relationships between scenario parameters and performance.

# Follow-up questions

A.Technical questions (metrics, ground-truth, and evaluation)

What specific performance metrics do you consider most relevant? (e.g., path length, time-to-goal, success/failure, localization error (RMSE), number of re-plans, collision counts, average cross-track error, cumulative rotation, etc.) — Please list any metrics you already use or prefer.
·  ·  What exactly do you define as a successful run? (reached goal within tolerance; no collisions; time limit; stable localization)
·  ·  Is ground-truth pose available for every timestamp in all runs (e.g., /tf map→base_link or a dedicated /ground_truth topic)? If not, how is ground-truth sampled?
·  ·  Which pose estimate should we treat as the robot's primary localization output for evaluation? (/amcl_pose, /odom, fused/localization node output)
·  ·  Are there annotated failures in the dataset (labels indicating failure reasons) or do we need to detect them automatically?
·  ·  Are there scenario-level meta labels available (e.g., complexity score, number of obstacles, obstacle sizes, noise level parameters), or should we derive them from filenames and config files?
·  ·  Could you provide the exact coordinate frames and TF tree convention used (frame names for map, odom, base_link, camera_link, imu_link)?
·  ·  Is the simulation deterministic for given seeds, or are there nondeterministic elements (e.g., random noise seeds not logged)?
·  ·  Are there time-synchronization guarantees between recorded topics (are headers and timestamps reliable across sensors and the rosbag)?
·  ·  Do you have any preferred threshold values for tolerances (e.g., distance-to-goal tolerance in meters, allowable orientation error in radians, collision/contact force thresholds)?


B. Data-related questions (availability, formats, provenance)



Could you provide a mapping of CSV columns to original ROS topics and message types? (A small data dictionary would be ideal.)

Are all runs recorded with the same ROS2 and Gazebo versions? If not, which runs differ and how?

Are the rosbag2 files complete, or are some runs missing topics due to recorder misconfiguration?

Are logs compressed/rotated or do any runs contain partial data due to early termination?

Are sensor models (LIDAR, IMU, camera) parameterized identically across scenarios, except for the noise parameters indicated in the filename, or do their intrinsic/extrinsic calibrations vary per scenario?

Does the dataset include the Nav2 and AMCL parameter files used in each run (e.g., amcl_params.yaml, nav2_params.yaml)?

Are the maps (PGM + YAML) aligned with the ground-truth map frame used in rosbag recordings?

What is the sampling frequency for each sensor topic (IMU, LIDAR, odom, amcl_pose) and does it vary between runs?

Are the scenario generator parameters (p1/p2, o4/o0, sd, n) fully described in a central manifest or only encoded in filenames?

Are there any privacy or IP constraints on sharing derived results or visuals beyond the research group?
